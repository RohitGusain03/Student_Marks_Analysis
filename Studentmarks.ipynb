```python
import numpy as np
```

```python
students = np.array(["Rahul","Aman","Priya","Suman","shalini"])

subjects = np.array(["English","Hindi","Maths","Geography","Science"])

marks = np. array([[85, 78, 92, 88, 76],
  [72, 80, 75, 79, 81],
  [90, 95, 89, 92, 94],
  [65, 70, 68, 72, 66],
  [88, 84, 91, 87, 90]
])
```

```python
print("Number of students:", marks.shape[0])
print("Number of subjects:", marks.shape[1])
print("shape:", marks.shape)
```

```python
total_marks = np.sum(marks, axis=1)

for student, total in zip(students, total_marks):
    print(student, ":", total)
```

```python
average_marks = np. mean(marks, axis = 1)

for student, average in zip(students, average_marks):
    print(f"{student}: {average:.2f}")
```

```python
top_index = np.argmax(average_marks)

print("Top Student:", students[top_index])
print("Average:", average_marks[top_index])
```

```python
subject_averages = np.mean(marks, axis = 0)

for subject, average in zip(subjects, subject_averages):
    print(f"{subject}: {average:.2f}")
```

```python
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
```

```python
high_performers = students[average_marks > 80]

print("students with average above 80:")

for student in high_performers: print(student)
```

```python
print("Overall average:", np.mean(marks))

print("Overall median:", np.median(marks))

print("Standard deviation:", np.std(marks))
```
